## Challanges In Building Agents Using Langchain##

**Glue code->When you Write the code which is not the library or framework code.**

**1)The langchain has no constructs or built features for non linear conditions so you write glue code which lead to hard maintanance**

**2)State is not there in langchain,memmory is there but that is conversational.While State is the dynamic thing that do keep changes untill reach his goal.we have multiple attributes for every scerio and everyone is considered and changes and is looked back by agent to current state.Tracking state is mandatory**

**State store values in key value pair while there no such a built in thing in langchain that can do this and you have to manually add dictionary and every time you have to change that dictionary manually  there is no auto update**

**The state Object can be accessed and can be made changes by any of the node,in this way the current node can access the state and look it to that what is next and what is done till now,That,s why langraph is called statefull**

**3)Event driven Execution -> langchain is sequential not event driven.Event driven mean to execute when a specific condition met etc.langraph is event driven**
