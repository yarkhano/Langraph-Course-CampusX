## Langraph ->Create flowchart and execute it, we have nodes and edges. It is orchestation framework(llm base apps framework) for building stateful,multistep, intelligent agents which has parallelism,memmory and resumability(start from where stopped) ##

## Langraph represent any LLM workflow in graph. ##

**Workflow->Series Of tasks to achieve end goal.**

## Common Workflows ##

**prompting-> Where multiple times your LLM is use,like report generation enhancing step by step.**

**Routing->LLM redirect to specific node or another LLM according to the type of task.like query is of billing,general or technical.**

**Parallelization-> You run multiple tasks parallel at same time.like we have to check multiple eligibility metrics and each is ran parallel.**

**Orchestator Workflow-> This is just like parallel workflow but here the rules are not predefined,but decided when input is given.**

**Evaluator Optimizer-> Where the task is evaluated and is done agin if not met evaluation criteria.like we have two llms one for generation second for evaluation(which reject + send feedback to the llm back or accept)**

## In langraph we represent the workflow in graph Where we have nodes and edges.Nodes is the functional part which is python function while edge is the decider what to do after execution of current node or function.edges can be sequential,paralle,branchied or loop.Every Node present a single task of workflow. ##

## State -> The shared memmory that is mutable and can by accessed by any node. The state of previous node is passed to next node and every node do changes accordingly and pass it onward to next node.State is key,value pair ##

## Reducers-> Control the state.either you want to wipe out the existig attribute and add new or append new to existing ##

**orphan node->not connected through edge**

**Super steps->because we have some time parallel execution that is why it is called super steps.**
